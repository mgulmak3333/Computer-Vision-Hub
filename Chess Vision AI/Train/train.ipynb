```python
import yaml

data_config = {
    'path': '/kaggle/input/datasets/imtkaggleteam/chess-pieces-detection-image-dataset/Chess_pieces',
    'train': 'train/images',
    'val': 'valid/images',
    'names': {
        0: 'bishop',
        1: 'black-bishop',
        2: 'black-king',
        3: 'black-knight',
        4: 'black-pawn',
        5: 'black-queen',
        6: 'black-rook',
        7: 'white-bishop',
        8: 'white-king',
        9: 'white-knight',
        10: 'white-pawn',
        11: 'white-queen',
        12: 'white-rook'
    }
}

with open('/kaggle/working/data.yaml', 'w') as f:
    yaml.dump(data_config, f, sort_keys=False)
```

```python
from ultralytics import YOLO

model = YOLO('/kaggle/input/models/musaglmak/yolov26n/pytorch/yolov26n/1/yolo26n.pt')

model.train(
    data = '/kaggle/working/data.yaml',
    epochs = 20,
    imgsz = 640,
    batch = 32,
    device = '0,1',
    workers = 4,
    name = 'chess_model_yolo'
)
```
